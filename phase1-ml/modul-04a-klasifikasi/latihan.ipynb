## Latihan 2

Sel berikut menyiapkan dua spesies iris dan memisahkan data latih dari data uji. Tidak ada unduhan. Bukan data asli perusahaan.

Setelah kamu melatih baseline, regresi logistik, dan pohon pendek (`max_depth=2`) pada data latih saja:

1. Berapa tebakan yang salah di data uji untuk baseline?
2. Berapa untuk regresi logistik?
3. Berapa untuk pohon pendek, di data latih dan di data uji?
4. Pohon tanpa batas kedalaman: apakah tebakan salah di data latih nol, dan apakah data ujinya lebih buruk daripada pohon pendek?


```python
# Satu ide: panggil pandas.
import pandas as pd
```

```python
# Satu ide: muat iris yang sudah terpasang dan buat tabel.
from sklearn.datasets import load_iris

iris = load_iris()
# feature_names adalah nama kolom bawaan dataset, bukan istilah baru.
tabel = pd.DataFrame(iris.data, columns=iris.feature_names)
tabel["spesies"] = iris.target_names[iris.target]
```

```python
# Satu ide: sisakan versicolor dan virginica.
dua = tabel.loc[tabel["spesies"] != "setosa"].copy()
```

```python
# Satu ide: pisahkan data latih dan data uji dengan SEED.
# test_size=0.25 berarti seperempat baris menjadi data uji.
# stratify menjaga proporsi golongan. random_state=SEED memakai angka tetap.
from sklearn.model_selection import train_test_split

ukuran = dua[list(iris.feature_names)]
golongan = dua["spesies"]
ukuran_latih, ukuran_uji, golongan_latih, golongan_uji = train_test_split(
    ukuran,
    golongan,
    test_size=0.25,
    random_state=SEED,
    stratify=golongan,
)
print("baris_latih", int(len(golongan_latih)))
print("baris_uji", int(len(golongan_uji)))
```
