Empat tinggi tanaman, dalam sentimeter: 5, 7, 6, dan 6.

1. Berapa jumlahnya?
2. Berapa `(5 + 7 + 6 + 6) / 4`?

Hitung dengan tangan, lalu jalankan sel di bawah dari atas ke bawah.


```python
# Satu ide: panggil NumPy, pustaka untuk hitungan pada array.
import numpy as np
```

```python
# Satu ide: simpan empat tinggi sebagai array, lalu lihat isinya.
tinggi_cm = np.array([5, 7, 6, 6])
tinggi_cm
```

```python
# Satu ide: jumlah seluruh isi array.
jumlah = int(tinggi_cm.sum())
print("jumlah", jumlah)
```

```python
# Satu ide: rata-rata = jumlah dibagi banyaknya data.
rata_rata = float(tinggi_cm.mean())
print("rata_rata", rata_rata)
```

```python
# Satu ide: panggil pandas, pustaka untuk tabel.
import pandas as pd
```

```python
# Satu ide: buat tabel kecil dengan nama kolom.
catatan = pd.DataFrame(
    {
        "tinggi_cm": [5, 7, 6, 6],
        "nama": ["mawar", "melati", "anggrek", "tulip"],
    }
)
catatan
```

```python
# Satu ide: rata-rata satu kolom. Kolom nama tidak ikut berubah.
rata_kolom = float(catatan["tinggi_cm"].mean())
print("rata_kolom", rata_kolom)
```

## Kasus nyata

Tabel bunga iris sudah ikut terpasang di scikit-learn. Sel di bawah tidak mengunduh apa pun. Ini bukan data asli perusahaan.

`sepal` artinya kelopak pelindung, `petal` artinya mahkota bunga. Keduanya diukur dalam sentimeter.


```python
# Satu ide: muat data iris dari berkas yang sudah terpasang.
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
# Satu ide: simpan angka iris sebagai DataFrame, dengan nama kolom dari dataset.
tabel = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
# Satu ide: mengubah nomor 0/1/2 menjadi nama jenis.
tabel["spesies"] = iris.target_names[iris.target]
```

```python
# Satu ide: cetak ukuran tabel.
print("baris", tabel.shape[0])
print("kolom", tabel.shape[1])
```

```python
# Satu ide: lihat beberapa baris pertama, seperti membuka halaman buku kas.
tabel.head()
```

```python
# Satu ide: rata-rata kolom yang sama, lewat DataFrame dan lewat array.
nama_kolom = "petal length (cm)"
rata_dari_tabel = float(tabel[nama_kolom].mean())
rata_dari_array = float(np.asarray(tabel[nama_kolom]).mean())
print("rata_dari_tabel", rata_dari_tabel)
print("rata_dari_array", rata_dari_array)
```

```python
# Satu ide: ambil lima baris. Indeks adalah nomor baris dari pandas, mulai dari 0.
# random_state memakai SEED supaya urutannya tetap.
cuplikan = tabel.sample(n=5, random_state=SEED)
print("indeks", cuplikan.index.tolist())
```

```python
# Satu ide: ulangi undian dengan SEED yang sama.
cuplikan_lagi = tabel.sample(n=5, random_state=SEED)
print("indeks_lagi", cuplikan_lagi.index.tolist())
print("sama", cuplikan.index.tolist() == cuplikan_lagi.index.tolist())
```
