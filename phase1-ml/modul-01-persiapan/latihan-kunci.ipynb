# Kunci latihan

Setiap sel jawaban mencetak hasilnya, lalu memeriksanya. Jalankan dari atas.


## Kunci latihan 1


```python
# Satu ide: jumlah dan rata-rata empat angka halaman.
import numpy as np

halaman = np.array([3, 5, 4, 8])
jumlah = int(halaman.sum())
rata_rata = float(halaman.mean())
print("jumlah", jumlah)
print("rata_rata", rata_rata)
assert jumlah == 20
assert rata_rata == 5.0
```

## Kunci latihan 2


```python
# Satu ide: rata-rata kolom halaman pada tabel.
import pandas as pd

buku = pd.DataFrame(
    {
        "hari": ["senin", "selasa", "rabu", "kamis"],
        "halaman": [3, 5, 4, 8],
    }
)
rata_kolom = float(buku["halaman"].mean())
print("rata_kolom", rata_kolom)
assert rata_kolom == 5.0
```

## Kunci latihan 3


```python
# Satu ide: ukuran tabel iris setelah kolom spesies ditambahkan.
from sklearn.datasets import load_iris

iris = load_iris()
tabel = pd.DataFrame(iris.data, columns=iris.feature_names)
# Mengubah nomor 0/1/2 menjadi nama jenis.
tabel["spesies"] = iris.target_names[iris.target]
print("baris", tabel.shape[0])
print("kolom", tabel.shape[1])
assert tabel.shape[0] == len(iris.data)
assert tabel.shape[1] == len(iris.feature_names) + 1
assert tabel.shape == (150, 5)
```

## Kunci latihan 4


```python
# Satu ide: dua undian dengan SEED yang sama memberi indeks yang sama.
pertama = tabel.sample(n=5, random_state=SEED)
kedua = tabel.sample(n=5, random_state=SEED)
print("indeks", pertama.index.tolist())
print("indeks_lagi", kedua.index.tolist())
print("sama", pertama.index.tolist() == kedua.index.tolist())
assert pertama.index.tolist() == kedua.index.tolist()
```
