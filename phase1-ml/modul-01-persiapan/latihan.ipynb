## Latihan 3

Sel berikut memuat tabel bunga iris yang sudah terpasang di scikit-learn, lalu menambah kolom jenis bunga. Tidak ada unduhan.

1. Ada berapa baris?
2. Ada berapa kolom setelah kolom spesies ditambahkan?


```python
# Satu ide: siapkan tabel. Ukuran baris dan kolom kamu hitung sendiri.
from sklearn.datasets import load_iris

iris = load_iris()
tabel = pd.DataFrame(iris.data, columns=iris.feature_names)
# Mengubah nomor 0/1/2 menjadi nama jenis.
tabel["spesies"] = iris.target_names[iris.target]
print(list(tabel.columns))
```

## Latihan 4

Indeks adalah nomor baris dari pandas, mulai dari 0.

Ambil lima baris dengan `random_state=SEED`, dua kali. Apakah kedua daftar indeks sama?


```python
# Satu ide: dua undian dengan SEED yang sama.
pertama = tabel.sample(n=5, random_state=SEED)
kedua = tabel.sample(n=5, random_state=SEED)
print("indeks", pertama.index.tolist())
print("indeks_lagi", kedua.index.tolist())
print("sama", pertama.index.tolist() == kedua.index.tolist())
```
