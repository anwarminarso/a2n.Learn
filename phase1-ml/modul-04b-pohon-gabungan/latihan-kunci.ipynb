# Kunci latihan

Setiap sel jawaban mencetak hasilnya, lalu memeriksanya. Jalankan dari atas.

Paket adalah bundel kode siap pakai. Pustaka adalah kumpulan fungsi di dalam paket.


## Kunci latihan 1


```python
# Satu ide: suara terbanyak dari tiga pemilah.
from collections import Counter

suara = ["matang", "mentah", "matang"]
hitung = Counter(suara)
terbanyak = hitung.most_common(1)[0][0]
print("hitung", dict(hitung))
print("terbanyak", terbanyak)
assert terbanyak == "matang"
assert hitung["matang"] == 2
```

## Kunci latihan 2


```python
# Satu ide: siapkan tabel anggur dan pecahan latih/uji yang sama dengan bahan latihan.
# test_size=0.25 berarti seperempat baris menjadi data uji.
# stratify menjaga proporsi golongan. random_state=SEED memakai angka tetap.
# fit artinya model belajar dari data latih. predict artinya model menebak golongan.
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

anggur = load_wine()
# feature_names adalah nama kolom bawaan dataset, bukan istilah baru.
tabel = pd.DataFrame(anggur.data, columns=anggur.feature_names)
tabel["golongan"] = anggur.target_names[anggur.target]
ukuran = tabel[list(anggur.feature_names)]
golongan = tabel["golongan"]
ukuran_latih, ukuran_uji, golongan_latih, golongan_uji = train_test_split(
    ukuran,
    golongan,
    test_size=0.25,
    random_state=SEED,
    stratify=golongan,
)
print("baris_uji", int(len(golongan_uji)))
```

```python
# Satu ide: satu pohon, hutan acak, dan gradient boosting pada data latih yang sama.
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    GradientBoostingClassifier,
    RandomForestClassifier,
)

pohon = DecisionTreeClassifier(random_state=SEED)
hutan = RandomForestClassifier(random_state=SEED)
boosting = GradientBoostingClassifier(random_state=SEED)
pohon.fit(ukuran_latih, golongan_latih)
hutan.fit(ukuran_latih, golongan_latih)
boosting.fit(ukuran_latih, golongan_latih)
```

```python
# Satu ide: banyak tebakan salah di data uji, dan nol di data latih bukan angka untuk anggur baru.
salah_uji_pohon = int(
    (pohon.predict(ukuran_uji) != golongan_uji.to_numpy()).sum()
)
salah_latih_pohon = int(
    (pohon.predict(ukuran_latih) != golongan_latih.to_numpy()).sum()
)
salah_uji_hutan = int(
    (hutan.predict(ukuran_uji) != golongan_uji.to_numpy()).sum()
)
salah_uji_boosting = int(
    (boosting.predict(ukuran_uji) != golongan_uji.to_numpy()).sum()
)
print("salah_latih_pohon", salah_latih_pohon)
print("salah_uji_pohon", salah_uji_pohon)
print("salah_uji_hutan", salah_uji_hutan)
print("salah_uji_boosting", salah_uji_boosting)
print("pakai_nol_latih_untuk_anggur_baru", False)
assert salah_latih_pohon == 0
assert salah_uji_pohon == 2
assert salah_uji_hutan == 0
assert salah_uji_boosting == 1
assert salah_uji_hutan < salah_uji_pohon
assert salah_uji_boosting < salah_uji_pohon
```

```python
# Satu ide: nama model dengan salah uji paling sedikit pada pecahan ini.
pasangan = {
    "pohon": salah_uji_pohon,
    "hutan_acak": salah_uji_hutan,
    "gradient_boosting": salah_uji_boosting,
}
print("paling_sedikit_salah_uji", min(pasangan, key=pasangan.get))
assert min(pasangan, key=pasangan.get) == "hutan_acak"
```
