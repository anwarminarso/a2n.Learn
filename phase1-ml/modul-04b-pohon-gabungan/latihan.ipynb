## Latihan 2

Sel berikut menyiapkan tabel anggur yang sudah terpasang, lalu memisahkan data latih dan data uji. Tidak ada unduhan. Bukan data asli perusahaan.

Setelah kamu melatih satu pohon, hutan acak, dan gradient boosting pada data latih saja:

1. Berapa tebakan salah di data uji untuk masing-masing?
2. Mana yang paling sedikit salah pada data uji ini?
3. Satu pohon salah nol kali di data latih. Apakah angka nol itu yang kamu pakai untuk anggur baru?


```python
# Satu ide: muat anggur yang sudah terpasang.
import pandas as pd
from sklearn.datasets import load_wine

anggur = load_wine()
# feature_names adalah nama kolom bawaan dataset, bukan istilah baru.
tabel = pd.DataFrame(anggur.data, columns=anggur.feature_names)
tabel["golongan"] = anggur.target_names[anggur.target]
```

```python
# Satu ide: ambil ukuran dan golongan anggur.
ukuran = tabel[list(anggur.feature_names)]
golongan = tabel["golongan"]
```

`test_size=0.25` berarti seperempat baris menjadi data uji. `stratify` menjaga proporsi golongan. `random_state=SEED` memakai angka tetap.


```python
# Satu ide: pisahkan data latih dan data uji dengan SEED.
from sklearn.model_selection import train_test_split
ukuran_latih, ukuran_uji, golongan_latih, golongan_uji = train_test_split(
    ukuran,
    golongan,
    test_size=0.25,
    random_state=SEED,
    stratify=golongan,
)
print("baris_latih", int(len(golongan_latih)))
print("baris_uji", int(len(golongan_uji)))
```
