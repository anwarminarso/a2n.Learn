## Kunci latihan 1


```python
# Satu ide: panggil NumPy.
import numpy as np
```

```python
# Satu ide: panggil KMeans.
from sklearn.cluster import KMeans
```

```python
# Satu ide: simpan enam titik latihan.
enam = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [8, 8],
    [8, 9],
    [20, 0],
])
```

```python
# Satu ide: dua kelompok pada enam titik.
label_kelompok = KMeans(n_clusters=2, n_init=10, random_state=SEED).fit_predict(enam)
print("label_kelompok", label_kelompok.tolist())
print("sama_dekat_nol", bool(label_kelompok[0] == label_kelompok[1] == label_kelompok[2]))
print("sama_dekat_delapan", bool(label_kelompok[3] == label_kelompok[4]))
print("aneh_ikut_delapan", bool(label_kelompok[5] == label_kelompok[3]))
assert bool(label_kelompok[0] == label_kelompok[1] == label_kelompok[2])
assert bool(label_kelompok[3] == label_kelompok[4])
assert bool(label_kelompok[5] == label_kelompok[3])
```

```python
# Satu ide: panggil IsolationForest.
from sklearn.ensemble import IsolationForest
```

```python
# Satu ide: tandai satu baris yang jauh.
hutan = IsolationForest(
    n_estimators=100,
    contamination=1 / 6,
    max_samples=len(enam),
    random_state=SEED,
)
label_aneh = hutan.fit_predict(enam)
indeks_aneh = int(np.where(label_aneh == -1)[0][0])
print("label_aneh", label_aneh.tolist())
print("indeks_aneh", indeks_aneh)
print("hanya_satu", bool(int(np.sum(label_aneh == -1)) == 1))
```

```python
# Satu ide: pastikan indeks baris aneh adalah 5.
assert indeks_aneh == 5
assert int(np.sum(label_aneh == -1)) == 1
```

## Kunci latihan 2


```python
# Satu ide: muat iris yang sudah terpasang.
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
# Satu ide: nama kolom angka, tanpa spesies.
nama = list(iris.feature_names)
print("nama", nama)
print("jumlah_kolom", len(nama))
print("ada_spesies", "spesies" in nama)
assert len(nama) == 4
assert "spesies" not in nama
```

## Kunci latihan 3


```python
# Satu ide: panggil Pipeline.
from sklearn.pipeline import Pipeline
```

```python
# Satu ide: panggil StandardScaler.
from sklearn.preprocessing import StandardScaler
```

```python
# Satu ide: panggil PCA.
from sklearn.decomposition import PCA
```

```python
# Satu ide: dua kolom baru, jumlah baris tetap.
pipa_ringkas = Pipeline([
    ("skala", StandardScaler()),
    ("ringkas", PCA(n_components=2, random_state=SEED)),
])
ringkas = pipa_ringkas.fit_transform(iris.data)
print("baris_awal", iris.data.shape[0])
print("baris_baru", ringkas.shape[0])
print("kolom_awal", iris.data.shape[1])
print("kolom_baru", ringkas.shape[1])
print("baris_tetap", iris.data.shape[0] == ringkas.shape[0])
```

```python
# Satu ide: pastikan 150 baris dan dua kolom.
assert iris.data.shape[0] == ringkas.shape[0] == 150
assert ringkas.shape[1] == 2
```
