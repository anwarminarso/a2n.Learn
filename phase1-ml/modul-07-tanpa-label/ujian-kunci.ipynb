```python
# Angka tetap. Setiap undian acak di notebook ini memakai SEED.
SEED = 42
```

# Kunci ujian tingkat advanced

Jalankan dari atas. Tiga baris `lulus_*` adalah hasil cek mini-proyek ini, bukan keputusan di luar notebook.

Pustaka adalah kumpulan fungsi siap pakai. Paket adalah program tambahan yang dipasang ke Python.


```python
# Satu ide: panggil NumPy.
import numpy as np
```

```python
# Satu ide: panggil generator angka.
from sklearn.datasets import make_blobs
```

```python
# Satu ide: panggil pemotong baris.
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: panggil Pipeline.
from sklearn.pipeline import Pipeline
```

```python
# Satu ide: panggil StandardScaler.
from sklearn.preprocessing import StandardScaler
```

```python
# Satu ide: panggil KMeans.
from sklearn.cluster import KMeans
```

```python
# Satu ide: buat angka, buang nomor gumpalan, potong data latih dan data uji.
X, _ = make_blobs(n_samples=120, centers=3, n_features=2, random_state=SEED)
X_latih, X_uji = train_test_split(X, test_size=0.25, random_state=SEED)
print("baris_latih", X_latih.shape[0])
print("baris_uji", X_uji.shape[0])
print("kolom", X_latih.shape[1])
```

```python
# Satu ide: Pipeline dari angka mentah sampai label kelompok, belajar hanya pada data latih.
pipa = Pipeline([
    ("skala", StandardScaler()),
    ("kelompok", KMeans(n_clusters=3, n_init=10, random_state=SEED)),
])
pipa.fit(X_latih)
print("nama_langkah", list(pipa.named_steps))
print("sampel_skala", int(pipa.named_steps["skala"].n_samples_seen_))
print("kolom_masukan", int(pipa.n_features_in_))
```

```python
# Satu ide: Pipeline kedua, SEED sama, label data uji harus sama.
pipa_lagi = Pipeline([
    ("skala", StandardScaler()),
    ("kelompok", KMeans(n_clusters=3, n_init=10, random_state=SEED)),
])
pipa_lagi.fit(X_latih)
sama = bool(pipa.predict(X_uji).tolist() == pipa_lagi.predict(X_uji).tolist())
print("sama", sama)
```

```python
# Satu ide: cara keliru, scaler melihat semua baris termasuk data uji.
scaler_keliru = StandardScaler()
scaler_keliru.fit(X)
print("sampel_skala_keliru", int(scaler_keliru.n_samples_seen_))
```

```python
# Satu ide: tiga syarat mini-proyek.
lulus_pipeline = list(pipa.named_steps) == ["skala", "kelompok"]
lulus_tanpa_bocor = int(pipa.named_steps["skala"].n_samples_seen_) == len(X_latih)
lulus_ulang = sama
print("lulus_pipeline", lulus_pipeline)
print("lulus_tanpa_bocor", lulus_tanpa_bocor)
print("lulus_ulang", lulus_ulang)
assert lulus_pipeline
assert lulus_tanpa_bocor
assert lulus_ulang
assert int(pipa.n_features_in_) == 2
assert int(scaler_keliru.n_samples_seen_) == len(X)
assert int(scaler_keliru.n_samples_seen_) != len(X_latih)
```
