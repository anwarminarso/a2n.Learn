Cerita lengkapnya ada di `README.md`. Di sini: lihat gambar dulu, hitung dengan tangan, baru jalankan sel dari atas.

Pustaka adalah kumpulan fungsi siap pakai. NumPy adalah pustaka untuk hitungan pada array.

Indeks adalah nomor baris, mulai dari 0.

## Coba sendiri dulu

```
y
9 |           D
8 |         C
2 | B
1 | A
  +------------- x
    1         8 9
```

A di (1, 1), B di (1, 2), C di (8, 8), D di (9, 8).

Kuadrat jarak = (selisih x)^2 + (selisih y)^2. Tanda ^2 di sini artinya dikali dirinya. Di kode, pangkat ditulis `**`.

1. Mana yang sekelompok: A dengan B, atau A dengan C?
2. Berapa kuadrat jarak A ke B, dan A ke C?

Hitung dulu. Jangan gulir ke angka cetak sebelum kamu punya jawaban sendiri.


````python
# Satu ide: panggil NumPy, pustaka untuk hitungan pada array.
import numpy as np
````

````python
# Satu ide: simpan empat titik pada gambar sebagai array.
titik = np.array([
    [1, 1],
    [1, 2],
    [8, 8],
    [9, 8],
])
````

````python
# Satu ide: kuadrat jarak A ke B dan A ke C.
selisih_ab = titik[0] - titik[1]
selisih_ac = titik[0] - titik[2]
kuadrat_dekat = int(np.sum(selisih_ab ** 2))
kuadrat_jauh = int(np.sum(selisih_ac ** 2))
akar_jauh = float(np.sqrt(kuadrat_jauh))
print("kuadrat_dekat", kuadrat_dekat)
print("kuadrat_jauh", kuadrat_jauh)
print("akar_jauh", akar_jauh)
````

KMeans adalah satu cara clustering: setiap titik ditaruh ke pusat kelompok terdekat, pusat digeser ke rata-rata anggotanya, lalu langkah itu diulang. Jumlah kelompok harus disebut dari awal. `n_init` adalah berapa kali pusat awal dicoba; kita tulis 10 supaya percobaan itu tidak mengikuti bawaan yang bisa berubah.

`fit` artinya langkah itu belajar dari baris yang diberikan. `random_state=SEED` mengunci undian pusat awal.


````python
# Satu ide: panggil KMeans dari paket scikit-learn.
from sklearn.cluster import KMeans
````

````python
# Satu ide: dua kelompok pada empat titik gambar.
model = KMeans(n_clusters=2, n_init=10, random_state=SEED)
label = model.fit_predict(titik)
print("label", label.tolist())
print("sama_AB", bool(label[0] == label[1]))
print("sama_CD", bool(label[2] == label[3]))
print("berbeda", bool(label[0] != label[2]))
assert int(model.n_clusters) == 2, "Empat titik memakai dua kelompok."
assert int(len(set(label.tolist()))) == 2, (
    "Hasil empat titik harus dua kelompok, bukan hanya hiperparameter."
)
assert label.tolist() == [0, 0, 1, 1], "Label empat titik harus sama dengan keluaran."
````

````python
# Satu ide: ulang dengan SEED yang sama.
model_lagi = KMeans(n_clusters=2, n_init=10, random_state=SEED)
label_lagi = model_lagi.fit_predict(titik)
print("sama", bool(label.tolist() == label_lagi.tolist()))
````

## Kasus nyata

Tabel bunga iris sudah ikut terpasang bersama scikit-learn. Sel di bawah tidak mengunduh apa pun. Ini bukan data asli perusahaan.

DataFrame, dari modul persiapan, adalah tabel di memori: setiap kolom punya nama, setiap baris satu catatan. Kita tidak menambah kolom spesies. Jumlah kelompok 2 kita pilih untuk dicoba, bukan dengan mengintip spesies.

Pada kasus ini tidak ada baris yang ditahan. Ini merangkum seluruh tabel, bukan ujian kebocoran. Ujian menahan sebagian baris.

Pipeline, topik modul rekayasa fitur, adalah urutan langkah yang diikat jadi satu: tiap langkah belajar hanya dari data yang diberikan ke `fit`. StandardScaler menggeser tiap kolom supaya rata-ratanya 0 dan sebarannya 1, supaya kolom berangka besar tidak menelan yang kecil.


````python
# Satu ide: panggil pandas, pustaka untuk tabel.
import pandas as pd
````

````python
# Satu ide: muat iris dari berkas yang sudah terpasang.
from sklearn.datasets import load_iris

iris = load_iris()
````

````python
# Satu ide: simpan empat ukuran bunga sebagai DataFrame, tanpa spesies.
tabel = pd.DataFrame(iris.data, columns=iris.feature_names)
````

````python
# Satu ide: lihat ukuran tabel.
print("baris", tabel.shape[0])
print("kolom", tabel.shape[1])
````

````python
# Satu ide: lihat nama kolom. Spesies tidak ada.
print("nama", list(tabel.columns))
````

````python
# Satu ide: panggil Pipeline.
from sklearn.pipeline import Pipeline
````

````python
# Satu ide: panggil StandardScaler.
from sklearn.preprocessing import StandardScaler
````

````python
# Satu ide: skala lalu clustering, pada empat kolom angka saja.
pipa_kelompok = Pipeline([
    ("skala", StandardScaler()),
    ("kelompok", KMeans(n_clusters=2, n_init=10, random_state=SEED)),
])
label_iris = pipa_kelompok.fit_predict(iris.data)
for nomor in sorted(set(label_iris.tolist())):
    print("ukuran", nomor, int((label_iris == nomor).sum()))
````

````python
# Satu ide: ukuran dua kelompok iris harus 100 dan 50.
kelompok = pipa_kelompok.named_steps["kelompok"]
ukuran = {
    nomor: int((label_iris == nomor).sum())
    for nomor in sorted(set(label_iris.tolist()))
}
assert ukuran == {0: 100, 1: 50}, "Ukuran kelompok iris harus 100 dan 50, sama dengan keluaran."
assert int(len(set(label_iris.tolist()))) == 2, "Hasil clustering iris harus dua kelompok."
assert int(len(set(label_iris.tolist()))) == int(kelompok.n_clusters), (
    "Banyak kelompok pada hasil harus sama dengan n_clusters yang terpakai."
)
````

PCA merangkum banyak kolom menjadi lebih sedikit kolom baru. Di scikit-learn, PCA menggeser tiap kolom ke rata-rata 0, tetapi tidak menyamakan sebaran. Karena itu scaler tetap di depan.

`explained_variance_ratio_` adalah bagian variasi yang tersimpan di tiap kolom baru.


````python
# Satu ide: panggil PCA.
from sklearn.decomposition import PCA
````

````python
# Satu ide: dua kolom baru, lalu bagian variasi yang tersimpan.
pipa_ringkas = Pipeline([
    ("skala", StandardScaler()),
    ("ringkas", PCA(n_components=2, random_state=SEED)),
])
ringkas = pipa_ringkas.fit_transform(iris.data)
langkah_pca = pipa_ringkas.named_steps["ringkas"]
rasio = langkah_pca.explained_variance_ratio_
print("variasi_1", float(rasio[0]))
print("variasi_2", float(rasio[1]))
print("variasi_tersimpan", float(rasio.sum()))
````

````python
# Satu ide: pastikan dua kolom dan 150 baris.
assert int(langkah_pca.n_components_) == 2, "Hasil PCA harus dua komponen."
assert int(ringkas.shape[1]) == int(langkah_pca.n_components_), (
    "Jumlah kolom hasil harus sama dengan n_components yang terpakai."
)
assert ringkas.shape[0] == 150, "PCA harus menjaga 150 baris, sama dengan keluaran."
````

````python
# Satu ide: bandingkan jumlah baris dan kolom, sebelum dan sesudah.
print("baris_awal", iris.data.shape[0])
print("baris_baru", ringkas.shape[0])
print("kolom_awal", iris.data.shape[1])
print("kolom_baru", ringkas.shape[1])
````

## Titik yang jauh

Enam titik: tiga dekat (0, 0), dua dekat (8, 8), satu jauh di (20, 0). Hitung dulu indeks titik yang jauh, mulai dari 0. Lalu duga: apakah clustering dengan dua kelompok akan menaruhnya sendiri?

IsolationForest adalah satu cara deteksi anomali: baris yang mudah dipisahkan dari mayoritas ditandai -1, baris biasa ditandai 1. `contamination` adalah perkiraan bagian baris yang aneh. Pada enam baris kita tulis 1/6, kira-kira satu baris. `max_samples` adalah banyak baris yang dipakai membangun satu pohon; kita samakan dengan jumlah baris tabel kecil ini.


````python
# Satu ide: simpan enam titik. Baris terakhir sengaja jauh.
enam = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [8, 8],
    [8, 9],
    [20, 0],
])
````

````python
# Satu ide: clustering dua kelompok pada enam titik.
label_kelompok = KMeans(n_clusters=2, n_init=10, random_state=SEED).fit_predict(enam)
print("label_kelompok", label_kelompok.tolist())
print("sama_dekat_nol", bool(label_kelompok[0] == label_kelompok[1] == label_kelompok[2]))
print("sama_dekat_delapan", bool(label_kelompok[3] == label_kelompok[4]))
print("aneh_ikut_delapan", bool(label_kelompok[5] == label_kelompok[3]))
````

````python
# Satu ide: tiga kelompok membuat titik yang jauh berdiri sendiri.
model_tiga = KMeans(n_clusters=3, n_init=10, random_state=SEED)
label_tiga = model_tiga.fit_predict(enam)
assert int(len(set(label_tiga.tolist()))) == 3, (
    "Hasil enam titik dengan n_clusters 3 harus tiga kelompok."
)
assert int(np.sum(label_tiga == label_tiga[5])) == 1, (
    "Dengan tiga kelompok, titik jauh harus berdiri sendiri."
)
````

````python
# Satu ide: panggil IsolationForest.
from sklearn.ensemble import IsolationForest
````

````python
# Satu ide: tandai baris yang jauh dari mayoritas.
hutan = IsolationForest(
    n_estimators=100,
    contamination=1 / 6,
    max_samples=len(enam),
    random_state=SEED,
)
label_aneh = hutan.fit_predict(enam)
# np.where mencari posisi yang nilainya -1.
indeks_aneh = int(np.where(label_aneh == -1)[0][0])
print("label_aneh", label_aneh.tolist())
print("indeks_aneh", indeks_aneh)
print("hanya_satu", bool(int(np.sum(label_aneh == -1)) == 1))
````

````python
# Satu ide: pastikan satu baris aneh pada indeks 5.
assert float(hutan.contamination) == 1 / 6, "Contamination harus 1/6."
assert indeks_aneh == 5, "Indeks baris aneh harus 5, sama dengan keluaran."
assert int(np.sum(label_aneh == -1)) == 1, "Jumlah baris aneh harus 1, sama dengan keluaran."
````
