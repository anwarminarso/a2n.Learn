# Ujian tingkat advanced

Modul ini menutup tingkat advanced. Kriteria pada roadmap: satu Pipeline lengkap dari data mentah sampai hasil, bebas data leakage, dan memberi hasil yang sama setiap kali diulang.

Pustaka adalah kumpulan fungsi siap pakai. Paket adalah program tambahan yang dipasang ke Python. Indeks adalah nomor baris, mulai dari 0.

Data latih dipakai model untuk belajar. Data uji disisihkan dan tidak dilihat saat belajar.

Pipeline, topik modul rekayasa fitur, adalah urutan langkah yang diikat jadi satu. Tiap langkah belajar hanya dari data yang diberikan ke `fit`.

## Mini-proyek

Bangun satu Pipeline pada angka di bawah:

1. Langkah `skala`: StandardScaler.
2. Langkah `kelompok`: KMeans dengan 3 kelompok, `n_init=10`, dan `random_state=SEED`.
3. `fit` hanya pada data latih. Jangan memasang scaler pada seluruh baris sebelum dipotong.
4. Nomor kelompok dari generator tidak boleh menjadi kolom.

`predict` artinya langkah yang sudah belajar memberi hasil pada baris, termasuk data uji.

Pertanyaan, kriteria lulus, dan kunci ada di `README.md` bagian ujian. Baca kunci setelah kamu mencoba. Solusi yang dicek ada di `ujian-kunci.ipynb`.

Generator `make_blobs` membuat angka di memori. Bukan unduhan. Bukan data asli perusahaan. Kita minta tiga gumpalan, lalu meminta tiga kelompok.


```python
# Satu ide: panggil NumPy.
import numpy as np
```

```python
# Satu ide: panggil generator angka dari paket scikit-learn.
from sklearn.datasets import make_blobs
```

```python
# Satu ide: panggil pemotong baris.
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: buat 120 baris, buang nomor gumpalan, lalu tahan seperempatnya.
X, _ = make_blobs(n_samples=120, centers=3, n_features=2, random_state=SEED)
X_latih, X_uji = train_test_split(X, test_size=0.25, random_state=SEED)
print("baris_latih", X_latih.shape[0])
print("baris_uji", X_uji.shape[0])
print("kolom", X_latih.shape[1])
```
