```python
# Satu ide: siapkan KMeans dengan SEED. Hasil kelompok ada di kunci.
from sklearn.cluster import KMeans

siap = KMeans(n_clusters=2, n_init=10, random_state=SEED)
print("siap", siap.random_state == SEED)
```
