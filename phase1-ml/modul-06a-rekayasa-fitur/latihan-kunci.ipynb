# Kunci latihan

Setiap sel jawaban mencetak hasilnya, lalu memeriksanya. Jalankan dari atas.


## Kunci latihan 1


```python
# Pustaka adalah paket kode siap pakai yang kita panggil.
# Satu ide: jumlah dan rata-rata empat gelas.
import numpy as np

gelas = np.array([100, 20, 80, 60])
jumlah = int(gelas.sum())
rata_rata = float(gelas.mean())
print("jumlah", jumlah)
print("rata_rata", rata_rata)
assert jumlah == 260
assert rata_rata == 65.0
```

## Kunci latihan 2


```python
# Encoding mengubah kategori menjadi angka. one-hot: satu kolom per kategori, 1 jika cocok.
# sparse_output=False meminta array biasa, yang menyimpan semua angka termasuk nol.
# Satu ide: encoding baris pertama.
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

suhu = pd.DataFrame({"suhu": ["panas", "dingin", "panas", "dingin"]})
pengkode = OneHotEncoder(sparse_output=False)
hasil = pengkode.fit_transform(suhu)
print("nama", pengkode.get_feature_names_out(["suhu"]).tolist())
print("baris_pertama", hasil[0].tolist())
assert pengkode.get_feature_names_out(["suhu"]).tolist() == [
    "suhu_dingin",
    "suhu_panas",
]
assert hasil[0].tolist() == [0.0, 1.0]
```

## Kunci latihan 3


```python
# fit belajar dari data latih. Tiga angka pertama saja. Angka 60 tidak ikut.
# Satu ide: rata-rata latih berbeda dari rata-rata seluruh angka.
from sklearn.preprocessing import StandardScaler

latih = gelas[:3].reshape(-1, 1)
semua = gelas.reshape(-1, 1)
penimbang = StandardScaler()
penimbang.fit(latih)
penimbang_salah = StandardScaler()
penimbang_salah.fit(semua)
print("rata_latih", float(penimbang.mean_[0]))
print("rata_semua", float(penimbang_salah.mean_[0]))
print("beda", float(penimbang.mean_[0]) != float(penimbang_salah.mean_[0]))
assert float(penimbang.mean_[0]) == float(np.mean(gelas[:3]))
assert float(penimbang.mean_[0]) != float(penimbang_salah.mean_[0])
```
