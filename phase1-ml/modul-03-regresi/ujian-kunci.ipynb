# Kunci ujian tingkat beginner

Sel ini menjalankan proyek mini dan memeriksa bahwa garis mengalahkan baseline pada data uji. Rubrik pertanyaan ada di `README.md`.

Pustaka adalah kumpulan fungsi siap pakai.


```python
# Satu ide: siapkan alat untuk proyek mini.
# n_samples banyak baris, n_features banyak petunjuk, noise: gangguan acak pada jawaban (derau).
import numpy as np
from sklearn.datasets import make_regression
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: buat angka proyek mini.
petunjuk, jawaban = make_regression(
    n_samples=80,
    n_features=1,
    n_informative=1,
    noise=15,
    random_state=SEED,
)
```

```python
# Satu ide: sisihkan data uji.
petunjuk_latih, petunjuk_uji, jawaban_latih, jawaban_uji = train_test_split(
    petunjuk, jawaban, test_size=0.25, random_state=SEED
)
print("ujian_baris_latih", int(len(jawaban_latih)))
print("ujian_baris_uji", int(len(jawaban_uji)))
assert len(jawaban_latih) == 60
assert len(jawaban_uji) == 20
```

```python
# Satu ide: garis dibanding baseline pada data uji.
baseline = DummyRegressor(strategy="mean")
baseline.fit(petunjuk_latih, jawaban_latih)
garis = LinearRegression()
garis.fit(petunjuk_latih, jawaban_latih)
mae_baseline = float(mean_absolute_error(jawaban_uji, baseline.predict(petunjuk_uji)))
mae_garis = float(mean_absolute_error(jawaban_uji, garis.predict(petunjuk_uji)))
print("ujian_mae_baseline", mae_baseline)
print("ujian_mae_garis", mae_garis)
print("ujian_lulus", mae_garis < mae_baseline)
assert mae_garis < mae_baseline
assert float(baseline.predict(petunjuk_uji)[0]) == float(jawaban_latih.mean())
```
