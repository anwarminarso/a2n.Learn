# Modul 3: Regresi

Cerita lengkapnya ada di `README.md`. Di sini: hitung dulu, baru jalankan kode.

Tiga perjalanan angkot yang catatannya terbuka. Jarak dalam kilometer: 1, 2, dan 3. Menitnya: 4, 6, dan 9. Garis yang dicoba: menit = 2 + 2 × jarak. Satu perjalanan di amplop: 4 km dan 12 menit.

1. Berapa tebakan garis itu pada jarak 1, 2, dan 3?
2. Berapa kuadrat selisihnya terhadap 4, 6, dan 9?
3. Berapa (0 + 0 + 1) / 3?
4. Berapa rata-rata 4, 6, dan 9?
5. Pada amplop, mana yang lebih dekat: tebakan garis, atau rata-rata itu?

Hitung dengan tangan, lalu jalankan sel dari atas ke bawah.


```python
# Satu ide: panggil NumPy. Pustaka adalah kumpulan fungsi siap pakai.
import numpy as np
```

```python
# Satu ide: simpan jarak dan menit yang terbuka.
jarak_latih = np.array([1, 2, 3], dtype=float)
menit_latih = np.array([4, 6, 9], dtype=float)
jarak_latih
```

```python
# Satu ide: tebakan garis yang dicoba, menit = 2 + 2 * jarak.
tebakan_coba = 2 + 2 * jarak_latih
print("tangan_tebakan", tebakan_coba.tolist())
```

```python
# Satu ide: kuadrat selisih. Selisih dikali dirinya sendiri.
kuadrat = (menit_latih - tebakan_coba) ** 2
print("tangan_kuadrat", kuadrat.tolist())
```

```python
# Satu ide: fungsi kerugian garis yang dicoba = rata-rata kuadrat itu.
print("tangan_fungsi_kerugian", float(kuadrat.mean()))
```

```python
# Satu ide: baseline adalah rata-rata menit data latih, seperti di modul 2.
print("tangan_baseline", float(menit_latih.mean()))
```

```python
# Satu ide: panggil fungsi rata-rata selisih mutlak dari modul 2.
from sklearn.metrics import mean_absolute_error
```

```python
# Satu ide: jarak pada amplop, garis coba dibanding baseline.
jarak_amplop = np.array([4.0])
menit_amplop = np.array([12.0])
tebakan_amplop_coba = 2 + 2 * jarak_amplop
mae_coba = float(mean_absolute_error(menit_amplop, tebakan_amplop_coba))
mae_base = float(mean_absolute_error(menit_amplop, np.full(1, menit_latih.mean())))
print("tangan_mae_garis_coba", mae_coba)
print("tangan_mae_baseline", mae_base)
print("tangan_coba_lebih_dekat", mae_coba < mae_base)
```

## Garis yang disusun pustaka

Garis di atas kita pilih dengan tangan. Berikutnya pustaka mencari garis regresi linear yang fungsi kerugiannya lebih kecil.

Angka awal adalah tebakan saat petunjuk bernilai nol. Angka pengali adalah tambahan tebakan tiap petunjuk bertambah satu. `fit` artinya menyusun garis dari data latih. `predict` artinya menghitung tebakan pada baris yang diberi. `reshape(-1, 1)` mengubah satu deret angka menjadi satu kolom, karena garis ini minta petunjuk berbentuk kolom.


```python
# Satu ide: panggil pembuat garis lurus di scikit-learn.
from sklearn.linear_model import LinearRegression
```

```python
# Satu ide: susun garis dari tiga baris terbuka, lalu cetak dua angkanya.
garis_tangan = LinearRegression()
garis_tangan.fit(jarak_latih.reshape(-1, 1), menit_latih)
print("tangan_awal", float(garis_tangan.intercept_))
print("tangan_pengali", float(garis_tangan.coef_[0]))
```

```python
# Satu ide: fungsi kerugian garis yang disusun pustaka, pada data latih.
tebakan_pas = garis_tangan.predict(jarak_latih.reshape(-1, 1))
kerugian_pas = float(np.mean((menit_latih - tebakan_pas) ** 2))
print("tangan_fungsi_kerugian_pas", kerugian_pas)
```

```python
# Satu ide: kerugian garis pustaka lebih kecil daripada garis yang dicoba.
print("tangan_pas_lebih_kecil", kerugian_pas < float(kuadrat.mean()))
```

```python
# Satu ide: jarak garis pustaka pada amplop, dibanding baseline.
tebakan_amplop_pas = garis_tangan.predict(jarak_amplop.reshape(-1, 1))
mae_pas = float(mean_absolute_error(menit_amplop, tebakan_amplop_pas))
print("tangan_mae_pas", mae_pas)
print("tangan_pas_mengalahkan_baseline", mae_pas < mae_base)
```

## Kasus nyata

Dataset diabetes sudah ikut terpasang di scikit-learn. Sel di bawah tidak mengunduh apa pun. Ini bukan data asli perusahaan.

Kita memakai satu petunjuk, kolom `bmi`, dan satu jawaban, kolom `target`. `bmi` pada dokumentasi adalah ukuran tubuh. Di tabel ini angkanya sudah dipusatkan dan diskalakan pembuat dataset, bukan kilogram yang kamu timbang di rumah. `target` adalah jawaban: ukuran seberapa jauh penyakit berkembang, satu tahun setelah pengukuran awal. Dokumentasinya tidak memberi satuan, jadi satuan tidak dikarang di sini.

Indeks adalah nomor baris pada tabel, mulai dari 0. Indeks baru dicetak di sel terakhir.


```python
# Satu ide: panggil pandas, pustaka untuk tabel, seperti di modul 1.
import pandas as pd
```

```python
# Satu ide: muat diabetes dari berkas yang sudah terpasang.
from sklearn.datasets import load_diabetes

bahan = load_diabetes(as_frame=True, scaled=True)
```

```python
# Satu ide: ambil satu petunjuk dan satu jawaban.
tabel = pd.DataFrame({"bmi": bahan.frame["bmi"], "target": bahan.frame["target"]})
tabel.shape
```

```python
# Satu ide: cetak ukuran tabel.
print("kasus_baris", int(tabel.shape[0]))
print("kasus_kolom", int(tabel.shape[1]))
```

```python
# Satu ide: panggil fungsi pemisah data latih dan data uji, seperti di modul 2.
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: seperempat baris menjadi data uji. random_state memakai SEED.
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("kasus_baris_latih", int(len(latih)))
print("kasus_baris_uji", int(len(uji)))
```

```python
# Satu ide: panggil baseline yang selalu menebak rata-rata.
from sklearn.dummy import DummyRegressor
```

```python
# Satu ide: baseline dari rata-rata jawaban data latih saja.
baseline = DummyRegressor(strategy="mean")
baseline.fit(latih[["bmi"]], latih["target"])
print("kasus_baseline", float(latih["target"].mean()))
```

```python
# Satu ide: susun garis dari data latih, lalu cetak angka awal dan angka pengali.
garis = LinearRegression()
garis.fit(latih[["bmi"]], latih["target"])
print("kasus_awal", float(garis.intercept_))
print("kasus_pengali", float(garis.coef_[0]))
```

```python
# Satu ide: rata-rata selisih mutlak pada data uji, baseline lalu garis.
mae_uji_baseline = float(mean_absolute_error(uji["target"], baseline.predict(uji[["bmi"]])))
mae_uji_garis = float(mean_absolute_error(uji["target"], garis.predict(uji[["bmi"]])))
print("kasus_mae_uji_baseline", mae_uji_baseline)
print("kasus_mae_uji_garis", mae_uji_garis)
print("kasus_garis_mengalahkan_baseline", mae_uji_garis < mae_uji_baseline)
```

```python
# Satu ide: jarak di data latih. Menyalin jawaban membuat jarak itu nol.
mae_latih_garis = float(mean_absolute_error(latih["target"], garis.predict(latih[["bmi"]])))
mae_latih_hafal = float(mean_absolute_error(latih["target"], latih["target"]))
print("kasus_mae_latih_garis", mae_latih_garis)
print("kasus_mae_latih_hafal", mae_latih_hafal)
```

## Garis yang menempel

Gambar di `README.md`: garis yang dipelintir menempel di setiap titik terbuka, termasuk titik yang kebetulan meleset. Modul ini tidak memasang garis yang dipelintir itu.

Nama polanya **overfitting**. Artinya: model terlalu menyesuaikan diri dengan data latih, termasuk kebetulan-kebetulannya, sehingga hasilnya bagus di data latih tetapi buruk di data baru. Kata model di kalimat itu bahasa biasa untuk cara menebak, sama seperti di modul 2, dan belum menjadi istilah baru.

Sel berikut membuat angka di komputer, bukan mengunduh. Ada 24 baris dan 20 petunjuk. Hanya satu petunjuk yang benar-benar dipakai untuk membuat jawaban; petunjuk lain kebetulan. Jawaban juga diberi gangguan acak. Karena petunjuk lebih banyak daripada baris data latih, garis lurus bisa menempel di data latih. `n_samples` adalah banyak baris. `n_features` adalah banyak petunjuk. `n_informative` adalah banyak petunjuk yang benar-benar dipakai membuat jawaban. `noise`: gangguan acak pada jawaban (derau). Geseran tetap adalah parameter `bias`.


```python
# Satu ide: panggil pembuat angka regresi. Ini bukan unduhan.
from sklearn.datasets import make_regression
```

```python
# Satu ide: buat angka di komputer. random_state memakai SEED.
petunjuk, jawaban = make_regression(
    n_samples=24,
    n_features=20,
    n_informative=1,
    noise=50,
    random_state=SEED,
)
print("lentur_petunjuk", int(petunjuk.shape[1]))
```

```python
# Satu ide: seperempat baris menjadi data uji.
petunjuk_latih, petunjuk_uji, jawaban_latih, jawaban_uji = train_test_split(
    petunjuk, jawaban, test_size=0.25, random_state=SEED
)
print("lentur_baris_latih", int(len(jawaban_latih)))
```

```python
# Satu ide: bandingkan jarak garis dan baseline pada angka yang dibuat itu.
garis_lentur = LinearRegression()
garis_lentur.fit(petunjuk_latih, jawaban_latih)
baseline_lentur = DummyRegressor(strategy="mean")
baseline_lentur.fit(petunjuk_latih, jawaban_latih)
lentur_mae_latih = float(mean_absolute_error(jawaban_latih, garis_lentur.predict(petunjuk_latih)))
lentur_mae_uji = float(mean_absolute_error(jawaban_uji, garis_lentur.predict(petunjuk_uji)))
lentur_mae_base = float(mean_absolute_error(jawaban_uji, baseline_lentur.predict(petunjuk_uji)))
print("lentur_mae_latih", lentur_mae_latih)
print("lentur_mae_uji", lentur_mae_uji)
print("lentur_mae_uji_baseline", lentur_mae_base)
```

```python
# Satu ide: data latih terlihat rapat, data uji lebih buruk daripada baseline.
print("lentur_latih_lebih_kecil", lentur_mae_latih < lentur_mae_uji)
print("lentur_uji_lebih_buruk", lentur_mae_uji > lentur_mae_base)
```
