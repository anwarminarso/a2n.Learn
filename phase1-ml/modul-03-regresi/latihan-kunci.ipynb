# Kunci latihan

Setiap sel jawaban mencetak hasilnya, lalu memeriksanya. Jalankan dari atas.

Pustaka adalah kumpulan fungsi siap pakai.


## Kunci latihan 1


```python
# Satu ide: fungsi kerugian garis coba pada data latih.
import numpy as np
from sklearn.metrics import mean_absolute_error

jarak_latih = np.array([1, 2, 4], dtype=float)
menit_latih = np.array([3, 5, 8], dtype=float)
jarak_uji = np.array([3.0])
menit_uji = np.array([10.0])
tebakan_latih = 1 + 2 * jarak_latih
kerugian = float(np.mean((menit_latih - tebakan_latih) ** 2))
print("kunci_fungsi_kerugian", kerugian)
assert kerugian == (0 + 0 + 1) / 3
```

```python
# Satu ide: jarak garis coba dan baseline pada data uji.
baseline = float(menit_latih.mean())
mae_garis = float(mean_absolute_error(menit_uji, 1 + 2 * jarak_uji))
mae_baseline = float(mean_absolute_error(menit_uji, np.full(1, baseline)))
print("kunci_baseline", baseline)
print("kunci_mae_garis", mae_garis)
print("kunci_mae_baseline", mae_baseline)
print("kunci_garis_lebih_dekat", mae_garis < mae_baseline)
assert mae_garis == 3.0
assert mae_garis < mae_baseline
```

## Kunci latihan 2


```python
# Satu ide: siapkan kolom bp dan pecahan latih/uji.
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

bahan = load_diabetes(as_frame=True, scaled=True)
tabel = pd.DataFrame({"bp": bahan.frame["bp"], "target": bahan.frame["target"]})
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("kunci_baris_latih", int(len(latih)))
print("kunci_baris_uji", int(len(uji)))
assert len(latih) == 331
assert len(uji) == 111
```

```python
# Satu ide: garis pada kolom bp, dinilai di data uji dengan rata-rata selisih mutlak.
baseline = DummyRegressor(strategy="mean")
baseline.fit(latih[["bp"]], latih["target"])
garis = LinearRegression()
garis.fit(latih[["bp"]], latih["target"])
nilai_baseline = float(latih["target"].mean())
mae_baseline = float(mean_absolute_error(uji["target"], baseline.predict(uji[["bp"]])))
mae_garis = float(mean_absolute_error(uji["target"], garis.predict(uji[["bp"]])))
print("kunci_baseline_bp", nilai_baseline)
print("kunci_mae_uji_baseline", mae_baseline)
print("kunci_mae_uji_garis", mae_garis)
print("kunci_garis_mengalahkan", mae_garis < mae_baseline)
assert nilai_baseline == float(latih["target"].mean())
assert mae_garis < mae_baseline
```
