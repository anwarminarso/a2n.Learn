## Latihan 2

Tabel di bawah memakai dataset Linnerud yang sudah terpasang. Bukan unduhan. Bukan data asli perusahaan.

Petunjuk: kolom `Jumps`. Jawaban: kolom `Waist`. Seperempat baris disisihkan dengan `random_state=SEED`.

1. Berapa baris data latih dan data uji?
2. Berapa rata-rata jawaban pada data latih saja?
3. Berapa rata-rata selisih mutlak baseline pada data uji?


```python
# Satu ide: siapkan tabel dan pemisahannya. Angka baseline belum dicetak.
import pandas as pd
from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split

bahan = load_linnerud(as_frame=True)
tabel = pd.DataFrame(
    {
        "Jumps": bahan.data["Jumps"],
        "Waist": bahan.target["Waist"],
    }
)
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
uji
```

## Latihan 3

Pemisahan di atas diulang sekali lagi dengan `random_state=SEED`. Apakah daftar indeks data uji sama?


```python
# Satu ide: dua pemisahan dengan SEED yang sama.
latih_lagi, uji_lagi = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("indeks_uji", uji.index.tolist())
print("indeks_uji_lagi", uji_lagi.index.tolist())
print("latihan_indeks_sama", uji.index.tolist() == uji_lagi.index.tolist())
```
