## Kunci latihan 1


```python
# Pustaka adalah kumpulan fungsi siap pakai yang dipanggil dari kode.
# Satu ide: panggil NumPy.
import numpy as np
```

```python
# Satu ide: panggil fungsi rata-rata selisih mutlak.
# mean_absolute_error adalah nama fungsi untuk hitungan yang sama, bukan istilah baru.
from sklearn.metrics import mean_absolute_error
```

```python
# Satu ide: rata-rata data latih.
menit_latih = np.array([4, 6, 8], dtype=float)
menit_uji = np.array([4, 10], dtype=float)
rata = float(menit_latih.mean())
print("kunci_rata", rata)
assert rata == 6.0
```

```python
# Satu ide: rata-rata selisih mutlak baseline pada data uji.
# np.full membuat array yang semua isinya angka yang sama.
tebakan = np.full(menit_uji.shape, rata)
mae = float(mean_absolute_error(menit_uji, tebakan))
print("kunci_mae", mae)
assert mae == 3.0
```

## Kunci latihan 2


```python
# Satu ide: panggil pandas, pustaka untuk tabel.
import pandas as pd
```

```python
# Satu ide: muat Linnerud dari berkas yang sudah terpasang.
# as_frame=True meminta data sudah berbentuk tabel.
from sklearn.datasets import load_linnerud

bahan = load_linnerud(as_frame=True)
```

```python
# Satu ide: ambil dua kolom menjadi satu tabel.
tabel = pd.DataFrame(
    {
        "Jumps": bahan.data["Jumps"],
        "Waist": bahan.target["Waist"],
    }
)
```

```python
# Satu ide: panggil fungsi yang memisahkan data latih dan data uji.
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: cetak banyak baris data latih dan data uji.
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("kunci_baris_latih", int(len(latih)))
print("kunci_baris_uji", int(len(uji)))
assert len(latih) == 15
assert len(uji) == 5
```

```python
# Satu ide: baseline adalah rata-rata jawaban data latih.
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy="mean")
baseline.fit(latih[["Jumps"]], latih["Waist"])
rata = float(baseline.predict(latih[["Jumps"]])[0])
print("kunci_baseline", rata)
assert baseline.strategy == "mean", "Baseline harus memakai strategi rata-rata."
assert rata == float(latih["Waist"].mean()), (
    "Prediksi baseline harus sama dengan rata-rata jawaban latih."
)
assert bool(np.all(baseline.predict(uji[["Jumps"]]) == rata)), (
    "Prediksi baseline harus konstan."
)
```

```python
# Satu ide: rata-rata selisih mutlak baseline pada data uji.
# np.full membuat array yang semua isinya angka yang sama.
# np.abs membuang tanda minus.
tebakan = np.full(len(uji), rata)
mae = float(mean_absolute_error(uji["Waist"], tebakan))
print("kunci_mae_uji", mae)
assert mae == float(np.mean(np.abs(uji["Waist"].to_numpy() - rata)))
```
