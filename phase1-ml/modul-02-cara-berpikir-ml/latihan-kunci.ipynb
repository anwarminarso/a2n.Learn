# Kunci latihan

Setiap sel jawaban mencetak hasilnya, lalu memeriksanya. Jalankan dari atas.


## Kunci latihan 1


```python
# Satu ide: rata-rata data latih, lalu rata-rata selisih mutlak pada data uji.
import numpy as np
from sklearn.metrics import mean_absolute_error

menit_latih = np.array([4, 6, 8], dtype=float)
menit_uji = np.array([4, 10], dtype=float)
rata = float(menit_latih.mean())
tebakan = np.full(menit_uji.shape, rata)
mae = float(mean_absolute_error(menit_uji, tebakan))
print("kunci_rata", rata)
print("kunci_mae", mae)
assert rata == 6.0
assert mae == 3.0
```

## Kunci latihan 2


```python
# Satu ide: baseline rata-rata pada kolom Waist, dinilai di data uji.
import pandas as pd
from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split

bahan = load_linnerud(as_frame=True)
tabel = pd.DataFrame(
    {
        "Jumps": bahan.data["Jumps"],
        "Waist": bahan.target["Waist"],
    }
)
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
rata = float(latih["Waist"].mean())
tebakan = np.full(len(uji), rata)
mae = float(mean_absolute_error(uji["Waist"], tebakan))
print("kunci_baris_latih", int(len(latih)))
print("kunci_baris_uji", int(len(uji)))
print("kunci_baseline", rata)
print("kunci_mae_uji", mae)
assert len(latih) == 15
assert len(uji) == 5
assert mae == float(np.mean(np.abs(uji["Waist"].to_numpy() - rata)))
assert float(latih["Waist"].mean()) == rata
```

```python
# Satu ide: dua pemisahan dengan SEED yang sama memberi indeks yang sama.
latih_lagi, uji_lagi = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("indeks_uji", uji.index.tolist())
print("indeks_uji_lagi", uji_lagi.index.tolist())
print("latihan_indeks_sama", uji.index.tolist() == uji_lagi.index.tolist())
assert uji.index.tolist() == uji_lagi.index.tolist()
```
