# Modul 2: Cara berpikir ML

Cerita lengkapnya ada di `README.md`. Di sini: hitung dulu, baru jalankan kode.

Tiga catatan menit yang terbuka: 8, 10, dan 12. Satu catatan di amplop: 16.

1. Berapa rata-rata 8, 10, dan 12?
2. Kalau setiap tebakan adalah rata-rata itu, berapa jaraknya terhadap 16?
3. Pada data latih, jaraknya 2, 0, dan 2. Berapa (2 + 0 + 2) / 3?

Hitung dengan tangan, lalu jalankan sel dari atas ke bawah.


```python
# Satu ide: panggil NumPy untuk array, seperti di modul 1.
import numpy as np
```

```python
# Satu ide: simpan tiga menit terbuka dan satu menit di amplop.
menit_latih = np.array([8, 10, 12])
menit_uji = np.array([16])
menit_latih
```

```python
# Satu ide: baseline cerita ini adalah rata-rata data latih.
rata = float(menit_latih.mean())
print("tangan_rata", rata)
```

```python
# Satu ide: jarak = selisih yang tanda minusnya dibuang.
jarak = [int(abs(nilai - rata)) for nilai in menit_latih]
print("tangan_jarak", jarak)
```

```python
# Satu ide: panggil fungsi rata-rata selisih mutlak di scikit-learn.
from sklearn.metrics import mean_absolute_error
```

```python
# Satu ide: rata-rata selisih mutlak baseline pada data latih.
tebakan_latih = np.full(menit_latih.shape, rata)
tangan_mae = float(mean_absolute_error(menit_latih, tebakan_latih))
print("tangan_mae", tangan_mae)
```

```python
# Satu ide: jarak baseline pada satu catatan yang disisihkan.
print("tangan_jarak_uji", int(abs(menit_uji[0] - rata)))
```

```python
# Satu ide: rata-rata selisih mutlak baseline pada data uji.
tebakan_uji = np.full(menit_uji.shape, rata)
print("tangan_mae_uji", float(mean_absolute_error(menit_uji, tebakan_uji)))
```

```python
# Satu ide: selisih yang masih bertanda bisa saling menghapus.
selisih_bertanda = menit_latih - rata
print("tangan_bertanda", float(selisih_bertanda.mean()))
```

```python
# Satu ide: menyalin jawaban data latih membuat jarak di data latih menjadi nol.
print("tangan_mae_hafal", float(mean_absolute_error(menit_latih, menit_latih)))
```

## Kasus nyata

Dataset Linnerud sudah ikut terpasang di scikit-learn. Sel di bawah tidak mengunduh apa pun. Ini bukan data asli perusahaan.

Dua puluh baris dari klub kebugaran. Kita memakai satu petunjuk, kolom `Situps`, dan satu jawaban, kolom `Weight`. Dokumentasinya tidak menyebut satuan `Weight`, jadi satuan tidak dikarang di sini.


```python
# Satu ide: panggil pandas untuk tabel, seperti di modul 1.
import pandas as pd
```

```python
# Satu ide: muat Linnerud dari berkas yang sudah terpasang, lalu ambil dua kolom.
from sklearn.datasets import load_linnerud

bahan = load_linnerud(as_frame=True)
tabel = pd.DataFrame(
    {
        "Situps": bahan.data["Situps"],
        "Weight": bahan.target["Weight"],
    }
)
tabel.head()
```

```python
# Satu ide: cetak ukuran tabel yang dipakai.
print("kasus_baris", int(tabel.shape[0]))
print("kasus_kolom", int(tabel.shape[1]))
```

```python
# Satu ide: panggil fungsi yang memisahkan data latih dan data uji.
from sklearn.model_selection import train_test_split
```

```python
# Satu ide: seperempat baris menjadi data uji. random_state memakai SEED.
latih, uji = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("kasus_baris_latih", int(len(latih)))
print("kasus_baris_uji", int(len(uji)))
```

```python
# Satu ide: panggil baseline yang selalu menebak rata-rata.
from sklearn.dummy import DummyRegressor
```

```python
# Satu ide: hitung rata-rata jawaban data latih saja.
baseline = DummyRegressor(strategy="mean")
baseline.fit(latih[["Situps"]], latih["Weight"])
nilai_baseline = float(latih["Weight"].mean())
print("kasus_baseline", nilai_baseline)
```

```python
# Satu ide: tebakan baseline pada data uji semuanya sama.
tebakan = baseline.predict(uji[["Situps"]])
print("kasus_tebakan", float(tebakan[0]))
print("kasus_tebakan_sama", bool(np.all(tebakan == tebakan[0])))
```

```python
# Satu ide: rata-rata selisih mutlak baseline pada data uji dan pada data latih.
mae_uji = float(mean_absolute_error(uji["Weight"], tebakan))
mae_latih = float(mean_absolute_error(latih["Weight"], baseline.predict(latih[["Situps"]])))
print("kasus_mae_uji", mae_uji)
print("kasus_mae_latih", mae_latih)
```

```python
# Satu ide: angka fungsi baseline sama dengan rata-rata yang dihitung tangan.
tebakan_tangan = np.full(len(uji), nilai_baseline)
mae_tangan = float(mean_absolute_error(uji["Weight"], tebakan_tangan))
print("kasus_mae_sama", mae_uji == mae_tangan and float(tebakan[0]) == nilai_baseline)
```

```python
# Satu ide: menyalin jawaban data latih membuat jarak di data latih menjadi nol.
print("kasus_mae_hafal", float(mean_absolute_error(latih["Weight"], latih["Weight"])))
```

```python
# Satu ide: ulangi pemisahan dengan SEED yang sama.
latih_lagi, uji_lagi = train_test_split(tabel, test_size=0.25, random_state=SEED)
print("kasus_indeks_uji", uji.index.tolist())
print("kasus_indeks_uji_lagi", uji_lagi.index.tolist())
print("kasus_sama", uji.index.tolist() == uji_lagi.index.tolist())
```
