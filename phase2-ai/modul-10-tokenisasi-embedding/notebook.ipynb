# Modul 10: Embedding dan tokenisasi

Cerita lengkapnya ada di `README.md`. Mode cepat tidak mengunduh model. Kalimatnya sintetis, ditulis di repo. Bukan data asli perusahaan.

Sebelum menjalankan kode: kalimat "kucing duduk di tikar" dan "kucing tidur di tikar" seharusnya lebih mirip daripada "kucing duduk di tikar" dan "kapal layar menyeberang laut". Jawab dulu, baru jalankan.


```python
# Satu ide: panggil numpy untuk vektor.
import numpy as np
```

```python
# Satu ide: token adalah kata yang dipisah spasi, lalu ditulis huruf kecil.
def tokenkan(kalimat):
    return kalimat.lower().split()

print("token", tokenkan("Kucing Duduk"))
```

```python
# Satu ide: embedding menaruh tiap token pada vektor yang tetap karena SEED.
rng = np.random.default_rng(SEED)
peta = {}

def vektor_kalimat(kalimat, ukuran=8):
    baris = []
    for tok in tokenkan(kalimat):
        if tok not in peta:
            peta[tok] = rng.normal(size=ukuran)
        baris.append(peta[tok])
    return np.mean(baris, axis=0)
```

```python
# Satu ide: kemiripan vektor memakai kosinus, lalu pasangan dekat harus unggul.
def kemiripan(a, b):
    return float(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)))

dekat_a = vektor_kalimat("kucing duduk di tikar")
dekat_b = vektor_kalimat("kucing tidur di tikar")
jauh = vektor_kalimat("kapal layar menyeberang laut")
skor_dekat = kemiripan(dekat_a, dekat_b)
skor_jauh = kemiripan(dekat_a, jauh)
print("skor_dekat", round(skor_dekat, 6))
print("skor_jauh", round(skor_jauh, 6))
print("selisih_skor", round(skor_dekat - skor_jauh, 6))
assert skor_dekat - skor_jauh >= 0.3
```
