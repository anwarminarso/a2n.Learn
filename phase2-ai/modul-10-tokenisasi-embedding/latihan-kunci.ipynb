# Kunci latihan modul 10


```python
# Satu ide: token kalimat latihan adalah tiga kata huruf kecil.
kalimat_latihan = "Burung Biru Terbang"
token_latihan = kalimat_latihan.lower().split()
print("token_latihan", token_latihan)
assert token_latihan == ["burung", "biru", "terbang"]
```

```python
# Satu ide: buat vektor kalimat dengan peta yang diundi memakai SEED.
import numpy as np

rng = np.random.default_rng(SEED)
peta = {}

def vektor_kalimat(kalimat, ukuran=8):
    baris = []
    for tok in kalimat.lower().split():
        if tok not in peta:
            peta[tok] = rng.normal(size=ukuran)
        baris.append(peta[tok])
    return np.mean(baris, axis=0)
```

```python
# Satu ide: pasangan yang berbagi kata harus lebih mirip, selisih minimal 0.3.
def kemiripan(a, b):
    return float(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)))

skor_dekat = kemiripan(vektor_kalimat("mobil merah cepat"), vektor_kalimat("mobil biru cepat"))
skor_jauh = kemiripan(vektor_kalimat("mobil merah cepat"), vektor_kalimat("gunung tinggi bersalju"))
print("skor_dekat_latihan", round(skor_dekat, 6))
print("skor_jauh_latihan", round(skor_jauh, 6))
assert skor_dekat - skor_jauh >= 0.3
```
