# Kunci latihan modul 9

Jawaban untuk tiga latihan. Tiap sel satu ide.


```python
# Satu ide: gradien y = 3 * x pada x = 5 adalah pengalinya.
x_latihan = 5.0
pengali = 3.0
gradien_latihan = pengali
print("gradien_latihan", gradien_latihan)
assert gradien_latihan == 3.0
assert x_latihan == 5.0
```

```python
# Satu ide: ReLU mengubah negatif menjadi nol dan meneruskan yang lain.
import numpy as np

angka_relu = np.array([-2.0, 0.0, 4.0])
hasil_relu = np.maximum(angka_relu, 0.0)
print("hasil_relu", hasil_relu.tolist())
assert hasil_relu.tolist() == [0.0, 0.0, 4.0]
```

```python
# Satu ide: selisih dua fungsi kerugian melewati ambang 0.15.
kerugian_awal_latihan = 0.25
kerugian_akhir_latihan = 0.10
turun_latihan = kerugian_awal_latihan - kerugian_akhir_latihan
print("turun_latihan", turun_latihan)
assert turun_latihan >= 0.15
```
