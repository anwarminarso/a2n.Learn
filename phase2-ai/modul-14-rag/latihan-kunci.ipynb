# Kunci latihan modul 14


```python
# Satu ide: cari folder modul, lalu muat rag untuk kunci.
from pathlib import Path
import importlib.util

di_sini = Path.cwd()
calon = [di_sini, *di_sini.parents]
akar = next(p for p in calon if (p / "phase2-ai" / "modul-14-rag" / "rag.py").is_file())
folder = akar / "phase2-ai" / "modul-14-rag"
spec = importlib.util.spec_from_file_location("rag14k", folder / "rag.py")
rag = importlib.util.module_from_spec(spec)
spec.loader.exec_module(rag)
```

```python
# Satu ide: dua pertanyaan latihan sama dengan yang di latihan.
pertanyaan_rak = "kode rak buku cerita"
pertanyaan_jebakan = "harga tiket kereta bandung"
```

```python
# Satu ide: bangun indeks yang sama, lalu jawab dua soal latihan.
dokumen = rag.baca_dokumen(folder / "dokumen")
muatan = rag.bangun_indeks(rag.buat_potongan(dokumen), SEED, folder / rag.FOLDER_INDEKS)
rak = rag.jawab(pertanyaan_rak, muatan)
jebakan = rag.jawab(pertanyaan_jebakan, muatan)
print("status_rak_latihan", rak["status"])
print("sumber_rak_latihan", rak["sumber"])
print("status_jebakan_latihan", jebakan["status"])
assert rak["sumber"] == "perpustakaan.txt"
assert jebakan["status"] == "tanpa_sumber"
```
