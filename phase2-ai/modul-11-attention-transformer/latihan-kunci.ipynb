```python
# Satu ide: jumlah baris bobot yang sah dekat ke 1.
baris_bobot = [0.2, 0.3, 0.5]
jumlah_baris = sum(baris_bobot)
print("jumlah_baris", round(jumlah_baris, 6))
assert abs(jumlah_baris - 1.0) <= 1e-6
```

```python
# Satu ide: baris yang jumlahnya jauh dari 1 tidak lolos ambang.
baris_salah = [0.2, 0.2, 0.2]
jumlah_salah = sum(baris_salah)
print("jumlah_salah", round(jumlah_salah, 6))
assert abs(jumlah_salah - 1.0) > 1e-6
```
